```python
import AlgebraTiles as at
```

```python
# ---------------------------
# Quick demos (render to screen and save transparent PNGs)
# ---------------------------

# Example 1: 4(x+3)
path1 = at.draw_area_model(
    h_segments=[at.Segment("var", 1, "x"), at.Segment("const", 3)],
    v_segments=[at.Segment("const", 4)],
    title="4(x + 3) as an Area Model",
    #save_path="/mnt/data/area_model_4_(x+3).png",
    transparent=True,
    fig_scale=1.1
)
```

```python
# Example 2: (x+2)(x+3)
path2 = at.draw_area_model(
    h_segments=[at.Segment("var", 1, "x"), at.Segment("const", 2)],
    v_segments=[at.Segment("var", 1, "x"), at.Segment("const", 3)],
    title="(x + 2)(x + 3) Area Model",
    #save_path="/mnt/data/area_model_(x+2)(x+3).png",
    transparent=True,
    fig_scale=1.1
)

print("Saved files:")
print(path1)
print(path2)
```
